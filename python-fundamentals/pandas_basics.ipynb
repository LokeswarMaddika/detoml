```python
# Create DataFrame with missing values
data_missing = {
    "name": ["Alice", "Bob", "Charlie", "David"],
    "age": [25, None, 35, 28],
    "salary": [70000, 85000, None, 75000]
}
```

```python
df2 = pd.DataFrame(data_missing)
print(df2)
```

```python
print(df2.isnull().sum())  # count missing values per column
```

```python
# 1. Drop rows with any missing values
df_dropped = df2.dropna()
print(df_dropped)
```

```python
# 2. Fill with a specific value
df_filled = df2.fillna(0)
print(df_filled)
```

```python
# 3. Fill with mean (most common for numerical columns)
df2["age"] = df2["age"].fillna(df2["age"].mean())
df2["salary"] = df2["salary"].fillna(df2["salary"].mean())
print(df2)
```

### Which method to use when:

dropna() — when missing data is random and you have enough rows to spare

fillna(0) — when 0 is a meaningful value (counts, flags)

fillna(mean) — most common for numerical columns, preserves distribution

fillna(median) — better when data has outliers (median is more robust)

```python
# Read the fraud detection dataset
df_fraud = pd.read_csv("/Users/lokeshmaddika/detoml/week-03-classical-ml/fraud-detection/creditcard.csv")
print(df_fraud.shape)
print(df_fraud.head())
print(df_fraud.isnull().sum().sum())  # total missing values
```

```python
print(df_fraud.isnull().sum().sum())  # total missing values
```

df_fraud.isnull().sum().sum() — two sums because:

First .sum() — sums missing values per column, returns a Series:
Time      0
V1        0
V2        0
...
Class     0


Second .sum() — sums all those column totals into one single number — total missing values across entire dataset.

Result: 0 — no missing values in the fraud dataset. Confirmed what we found in the project.

One .sum() = missing per column
Two .sum() = total missing across all columns

```python
# Fraud vs legitimate transaction analysis
print(df_fraud.groupby("Class")["Amount"].agg(["mean", "max", "min"]))
```

Your intuition was right — fraud transactions average $122 vs $88 for legitimate. But notice the max: legitimate transactions go up to $25,691 while fraud maxes at $2,125.

What this tells us:

Fraudsters tend to make medium-sized transactions — not too small (suspicious), not too large (triggers alerts)
The extremely large transactions are actually legitimate
